```python
# Q21. Find the average age of sailors for each rating level that has at least two sailors.
```

```python
query20="""
select S.rating, avg(S.age) as Average_age from Sailors S group by S.rating having count(*)>1
"""
result=pd.read_sql_query(query20,conn)
result
```

```python
# Q22.  Find the average age of sailors who is eligible to vote (i.e., at least 18years old)
#  for each rating level that has at least two such sailors
```

```python
query22="""
select S.rating,  avg(S.age) as Average_age from Sailors S where S.age>=18 group by S.rating having count(*)>1
"""
result=pd.read_sql_query(query22,conn)
result
```

```python
# Doubt. Q23. Find those ratings for which average age of sailor is minimum over all ratings.
```

```python
query23="""
SELECT rating, avgage
FROM (
    SELECT S.rating, AVG(S.age) AS avgage
    FROM Sailors S
    GROUP BY S.rating
)
WHERE avgage = (
    SELECT MIN(avgage)
    FROM (
        SELECT AVG(S.age) AS avgage
        FROM Sailors S
        GROUP BY S.rating
    )
);
"""
result=pd.read_sql_query(query23,conn)
result
```
